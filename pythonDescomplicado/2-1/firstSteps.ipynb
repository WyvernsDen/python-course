```python
num1 = int(input("Input the first number: "))
num2 = int(input("Input the second number: "))

sum = num1 + num2
print("The sum of {1} and {2} is equal to: {0}".format(sum, num1, num2))
# .format formats according to the sequence of the numbers in {}
# 0 is the first, so sum is 0, 1 is second, so num1 is 1, 2 is the last, so num2 is 2
```
