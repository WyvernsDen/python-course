```python
num1 = int(input("Type in the first number: "))
num2 = int(input("Type in the second number: "))

if num1 > num2:
    print(f"{num1} is larger than {num2}.")
    print(f"The difference between {num1} and {num2} is {num1 - num2}.")
    print(f"The sum of {num1} and {num2} is {num1 + num2}.")
elif num1 < num2:
    print(f"{num2} is larger than {num1}.")
    print(f"The difference between {num2} and {num1} is {num2 - num1}.")
    print(f"The sum of {num2} and {num1} is {num2 + num1}.")
elif num1 == num2:
    print(f"{num1} is equal to {num2}.")
else:
    print("Undetermined.")

print("Thank you!")
```

```python
grade1 = float(input("Type in the grade for your first test: "))
grade2 = float(input("Type in the grade for your Second test: "))
avg = (grade1 + grade2) / 2.0

if avg >= 7:
    print(f"Congratulations! You are passing with an average of {avg}!")
else:
    print(f"Unfortunately you are not passing because of your average of {avg}")
```
