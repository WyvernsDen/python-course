```python
num = 6/7
print(num)
```

```python
print('Resultado: %f' % num)
# ^ shows float point with 6 decimal places
print('Resultado: %.2f' % num)
# ^ formats with 2 decimal places
print('Resultado: %8.2f' % num)
# ^ formats with 2 decimal places and an 8 character limit,
# remaining places are filled with spaces to the left
print('Resultado: %d' % num)
# ^ formats as int
print('Resultado: %s' % num)
# ^ formats as str
```

```python
num2 = 10
t1 = "Hello"
t2 = "World"

print(f'Resultado: {num:f}')
# ^ shows default 6 decimal places
print(f'Resultado: {num:.2f}')
# ^ formats with 2 decimal places
print(f'Resultado: {num:8.2f}')
# ^ formats with 2 decimal places and an 8 character limit,
# remaining places filled with spaces to the left
print(f'Resultado: {num2:8d}')
# ^ formats with a minimum of 8 characters to the left
# fills places to the left with spaces if needed
print(f'{t1:10}{t2:10}')
# ^ formats wth a minimum of 10 characters, no alignment
print(f'{t1:^10}{t2:^10}')
# ^ formats with a minimum of 10 characters, centered
print(f'{t1:>10}{t2:>10}')
# ^ formats with a minimum of 10 characters, aligned to right side
```

```python
print('Resultado: {0:f}'.format(num))
print('Resultado: {0:.2f}'.format(num))
print('Resultado: {0:8.2f}'.format(num))
print('Resultado: {0:8d}'.format(num2))
print('{0:10}{1:10}'.format(t1, t2))
print('{0:^10}{1:^10}'.format(t1, t2))
print('{0:>10}{1:>10}'.format(t1, t2))
# ^ you can use .format as well, looks much messier
```
